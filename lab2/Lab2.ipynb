## Завдання 6. Складна умова: Latte та сума більше 30
Дві умови поєднуємо оператором `&`, кожну беремо в дужки (оператор `and` тут не працює, бо порівнює цілі об'єкти, а не окремі елементи).

```python
latte_high = df[(df['coffee_name'] == 'Latte') & (df['money'] > 30)]
print('Усього рядків:', len(latte_high))
latte_high.head(10)
```

## Завдання 7. Сортування за `datetime`
Час записано у форматі `РРРР-ММ-ДД ГГ:ХХ:СС`, тому текстове сортування збігається з хронологічним. Метод `sort_values()` повертає новий об'єкт і не змінює `df`.

```python
sorted_df = df.sort_values('datetime')
sorted_df.head(3)
```

**Зауваження:** таблиця вже була впорядкована за часом, тому результат збігається з `df.head(3)`.

## ★ Додаткове завдання. Функція підрахунку за категоріальною колонкою
Функція рахує кількість записів у кожній групі через `value_counts()` і сортує результат за спаданням через `sort_values()`.

```python
def count_by_category(data, column):
    """Повертає DataFrame: категорії колонки та кількість записів, за спаданням."""
    result = (
        data[column]
        .value_counts()
        .sort_values(ascending=False)
        .reset_index()
    )
    result.columns = [column, 'count']
    return result

coffee_counts = count_by_category(df, 'coffee_name')
coffee_counts
```

Зберігаємо результат у CSV без колонки індексів.

```python
coffee_counts.to_csv('/content/drive/MyDrive/coffee_counts.csv', index=False)
print('Файл збережено!')
```
