## Завдання 4. Одновимірний масив NumPy
Створюємо масив із перших 10 парних чисел за допомогою `np.arange(2, 21, 2)` та обчислюємо агрегатні показники.

```python
arr = np.arange(2, 21, 2)
print("Масив:", arr)
print("Сума:", arr.sum())
print("Середнє:", arr.mean())
print("Мінімум:", arr.min())
print("Максимум:", arr.max())
print("Масив, помножений на 3:", arr * 3)
```

## Завдання 5. Двовимірний масив NumPy 3×3
Масив заповнено випадковими цілими від 1 до 10 (верхня межа `randint` не входить, тому вказано 11). Параметр `axis=0` дає суми за стовпцями, `axis=1` — за рядками.

```python
m = np.random.randint(1, 11, size=(3, 3))
print(m)
print("Суми за стовпцями (axis=0):", m.sum(axis=0))
print("Суми за рядками (axis=1):", m.sum(axis=1))
```
